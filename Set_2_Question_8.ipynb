```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files
import matplotlib.pyplot as plt

image = cv2.imread('/content/dog.png')

image = cv2.resize(image, (256, 256))

# Apply Laplacian filter (edge detection)
laplacian = cv2.Laplacian(image, cv2.CV_64F)
laplacian = np.clip(laplacian, 0, 255).astype(np.uint8)

# Sharpen by subtracting Laplacian from original image
sharpened = cv2.subtract(image, laplacian)

# Display results side by side
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(laplacian, cv2.COLOR_BGR2RGB))
plt.title('Laplacian (Edges)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title('Sharpened Image')
plt.axis('off')

plt.show()
```
