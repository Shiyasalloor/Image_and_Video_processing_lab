```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('/content/dog.png', cv2.IMREAD_GRAYSCALE)
img = cv2.resize(img, (256, 256))
```

```python
horizontal_mask = np.array([[-1, -1, -1],
                            [ 2,  2,  2],
                            [-1, -1, -1]])

vertical_mask = np.array([[-1,  2, -1],
                          [-1,  2, -1],
                          [-1,  2, -1]])

diagonal_mask = np.array([[ 2, -1, -1],
                          [-1,  2, -1],
                          [-1, -1,  2]])
```

```python
horizontal_lines = cv2.filter2D(img, -1, horizontal_mask)
vertical_lines   = cv2.filter2D(img, -1, vertical_mask)
diagonal_lines   = cv2.filter2D(img, -1, diagonal_mask)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(horizontal_lines, cmap='gray')
plt.title('Horizontal Lines Detected')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(vertical_lines, cmap='gray')
plt.title('Vertical Lines Detected')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(diagonal_lines, cmap='gray')
plt.title('Diagonal Lines Detected')
plt.axis('off')

plt.tight_layout()
plt.show()
```
